# Week 01 — Basic Data Exploration and Cleaning using Pandas

**Dataset:** Ecommerce Shopping Dataset ([Kaggle](https://www.kaggle.com/datasets/anvitkumar/shopping-dataset))  
**Objective:** Load, explore, clean, and save a CSV dataset using Pandas.

## Step 0 — Create Combined Dataset from Individual CSVs

> **Run this only once.** The downloaded Kaggle dataset contains one CSV per product category.
> This step merges all of them into a single `Combined_dataset.csv`.
> If you already have `Combined_dataset.csv`, skip to Step 1.

```python
import os
import glob
import pandas as pd
import numpy as np

# Path to the folder containing all individual category CSVs
CSV_FOLDER = 'raw_data'

# Find all CSV files except any existing combined file
all_csv_files = glob.glob(os.path.join(CSV_FOLDER, '*.csv'))
all_csv_files = [f for f in all_csv_files if 'Combined_dataset' not in f and 'cleaned' not in f]

print(f'Found {len(all_csv_files)} category CSV files:')
for f in sorted(all_csv_files):
    print(' ', os.path.basename(f))
```

```python
dfs = []

for filepath in sorted(all_csv_files):
    category_name = os.path.basename(filepath).replace('.csv', '')
    try:
        temp_df = pd.read_csv(filepath)
        temp_df['category'] = category_name   # tag each row with its source category
        dfs.append(temp_df)
    except Exception as e:
        print(f'  Skipped {category_name}: {e}')

# Merge all into one DataFrame
combined_df = pd.concat(dfs, ignore_index=True)

# Save
combined_df.to_csv('Combined_dataset.csv', index=False)

print(f'\nCombined dataset created!')
print(f'Total rows   : {len(combined_df):,}')
print(f'Total columns: {combined_df.shape[1]}')
print(f'Categories   : {combined_df["category"].nunique()}')
```

## Step 1 — Load the Combined Dataset

```python
df = pd.read_csv('Combined_dataset.csv')

print('Dataset loaded successfully!')
print(f'Shape: {df.shape}')
```

## Step 2 — Explore the Data

```python
# First and last 5 rows
print('--- HEAD ---')
display(df.head())
print('--- TAIL ---')
display(df.tail())
```

```python
# Shape and column names
print('Shape:', df.shape)
print('\nColumns:', df.columns.tolist())
```

```python
# Data types
df.dtypes
```

```python
# Basic statistics for numeric columns
df.describe()
```

```python
# Category breakdown — how many products per category?
category_counts = df['category'].value_counts()
print(f'Total categories: {len(category_counts)}')
print('\nTop 10 categories by product count:')
print(category_counts.head(10).to_string())
```

## Step 3 — Handle Missing Values

```python
# Check missing values per column
missing = df.isnull().sum()
missing_pct = (missing / len(df)) * 100

missing_df = pd.DataFrame({'Missing Count': missing, 'Missing %': missing_pct})
missing_df[missing_df['Missing Count'] > 0].sort_values('Missing %', ascending=False)
```

```python
# Drop columns with more than 50% missing values
cols_to_drop = missing_pct[missing_pct > 50].index.tolist()
print('Dropping columns with >50% missing:', cols_to_drop)
df.drop(columns=cols_to_drop, inplace=True)
```

```python
# Fill missing numeric values with column median
numeric_cols = df.select_dtypes(include=[np.number]).columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].median())

# Fill missing text values with 'Unknown'
categorical_cols = df.select_dtypes(include=['object', 'str']).columns
df[categorical_cols] = df[categorical_cols].fillna('Unknown')

print('Missing values after handling:', df.isnull().sum().sum())
```

## Step 4 — Basic Operations (Filter & Select)

```python
# Select key columns (only keep ones that exist)
key_cols = ['product_id', 'title', 'category', 'initial_price', 'final_price',
            'discount', 'rating', 'ratings_count', 'seller_name']
key_cols = [col for col in key_cols if col in df.columns]

df_filtered = df[key_cols].copy()
print('Working with columns:', key_cols)
df_filtered.head()
```

```python
# Filter: products with rating >= 4.0
if 'rating' in df_filtered.columns:
    high_rated = df_filtered[df_filtered['rating'] >= 4.0]
    print(f'Products with rating >= 4.0: {len(high_rated):,}')
    display(high_rated.head())
```

```python
# Filter: products with discount > 20%
if 'discount' in df_filtered.columns:
    discounted = df_filtered[df_filtered['discount'] > 20]
    print(f'Products with >20% discount: {len(discounted):,}')
    display(discounted.head())
```

## Step 5 — Remove Duplicates

```python
before = len(df_filtered)

# 1. Remove fully identical rows
df_filtered = df_filtered.drop_duplicates()
after_full = len(df_filtered)

# 2. Remove rows with duplicate product titles (keep first occurrence)
if 'title' in df_filtered.columns:
    df_filtered = df_filtered.drop_duplicates(subset='title', keep='first')
after_title = len(df_filtered)

print(f'Rows before          : {before:,}')
print(f'After full dedup     : {after_full:,}  (removed {before - after_full} exact duplicates)')
print(f'After title dedup    : {after_title:,}  (removed {after_full - after_title} duplicate titles)')
```

## Step 6 — Create Derived Column: `total_amount`

```python
# total_amount = final_price * ratings_count  (a rough proxy for product demand/revenue)
if 'final_price' in df_filtered.columns and 'ratings_count' in df_filtered.columns:
    df_filtered['total_amount'] = df_filtered['final_price'] * df_filtered['ratings_count']
    print('Derived column total_amount created.')
    display(df_filtered[['title', 'final_price', 'ratings_count', 'total_amount']].head(10))
```

## Step 7 — Save Cleaned Dataset

```python
df_filtered.to_csv('ecommerce_cleaned.csv', index=False)

print('Saved: ecommerce_cleaned.csv')
print(f'Final shape: {df_filtered.shape}')
```
